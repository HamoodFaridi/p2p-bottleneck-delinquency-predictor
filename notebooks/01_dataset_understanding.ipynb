# Data Understanding Conclusion

The initial dataset contains 50,000 invoice records and 19 columns.

The `isOpen` field was determined not to be a suitable ML target because it is directly associated with the presence or absence of `clear_date`.

For the 40,000 cleared invoices, payment behavior can be measured using the difference between `clear_date` and `due_in_date`.

The following target definitions have been established:

- `days_late = clear_date - due_in_date`
- `is_delayed = 1` when `days_late > 0`
- `is_delayed = 0` when `days_late <= 0`

The historical data contains 33.528% delayed invoices and 66.472% invoices that were paid on or before the due date.

Negative `days_late` values represent early payments and are considered valid business outcomes.

The `document_create_date.1` column matches `posting_date` for 99.988% of records and will be investigated during data cleaning as a potential duplicate/extraction artifact.

The dataset contains 10,000 currently open invoices for which the final payment outcome is not yet known. These records will not be used to calculate historical delinquency outcomes but may eventually be used as the population for production prediction.

The next phase will focus on data validation, standardizing date fields, handling missing values, identifying redundant columns, and preparing a clean dataset for target construction and feature engineering.

```python
def investigate_duplicate_records(
        df: pd.DataFrame
) -> None:
    """
    Investigate exact duplicate records and identify 
    whether duplicate invoice records contain differences
    """

    print("\n--- Duplicate Record Investigation ---")

    duplicate_rows = df[
        df.duplicated(
            keep=False
        )
    ].copy()

    print("Exact duplicated rows:", len(duplicate_rows))

    print("Duplicated groups:",duplicate_rows.duplicated().sum())

    # Investigate duplicates based on Invoice IDs
    if "invoice_id" in df.columns:
        invoice_duplicates = df[
            df["invoice_id"].duplicated(
                keep=False
            )
            & df["invoice_id"].notna()
        ].copy()

        print("\nRows with duplicated invoice_id:", len(invoice_duplicates))

        # Remove invoice_id and doc_id from comparison
        # to determine whether the remaining business
        # attributes are identical.

        comparison_columns = [
            column for column in invoice_duplicates.columns
            if column not in [
                "invoice_id",
                "doc_id",

            ]
        ] 

        business_duplicates = (
            invoice_duplicates
            .duplicated(
                subset = comparison_columns,
                keep = False,
            )
        )

        print("Duplicated invoice records with identical business attributes:", business_duplicates.sum(),)

        print("Duplicated invoice records with identical business attributes:", (~business_duplicates).sum(),)
```
